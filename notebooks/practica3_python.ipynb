```python
## Ejercicio 1 — Catálogo de productos como lista
```

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Total elementos y total productos (cada producto ocupa 2 posiciones: nombre y precio) dd
total_elementos = len(catalogo)
total_productos = total_elementos // 2
print(f"Total elementos: {total_elementos} -> Dedujimos {total_productos} productos.")

# 2. Slicing con paso: nombres (pares desde 0) y precios (impares desde 1)
nombres = catalogo[0::2]   
precios = catalogo[1::2]
print("Nombres:", nombres)
print("Precios:", precios)

# 3. Precio del Robot Aspirador usando index
pos_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[pos_robot + 1]
print(f"Precio del Robot Aspirador (índice {pos_robot + 1}): {precio_robot} €")

# 4. Tres últimos productos usando índices negativos
ultimos_tres_nombres = nombres[-3:]
print("Últimos tres productos:", ultimos_tres_nombres)

# 5. Precio medio del catálogo con sum(), len() y round()
precio_medio = round(sum(precios) / len(precios), 2)
print("Precio medio del catálogo:", precio_medio, "€")

# Comprobación requerida
print("Comprobación len(nombres):", len(nombres))
print("Comprobación sum(precios):", round(sum(precios), 2))
```

```python
## Ejercicio 5 — Funciones propias para el cálculo de importes
```

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Calcula el importe neto aplicando descuento porcentual."""
    bruto = unidades * precio_unitario
    neto = bruto * (1 - descuento_pct / 100)
    return round(neto, 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve tupla con (importe_bruto, ahorro, importe_neto)."""
    bruto = round(unidades * precio_unitario, 2)
    neto = round(bruto * (1 - descuento_pct / 100), 2)
    ahorro = round(bruto - neto, 2)
    return (bruto, ahorro, neto)

# Desempaquetar con argumentos nominales
bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print(f"Resumen pedido: Bruto={bruto} €, Ahorro={ahorro} €, Neto={neto} €")

# Documentación con help
help(calcular_importe)

pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = [calcular_importe(u, p, d) for u, p, d in pedidos]
print("Importes calculados:", importes)
print(f"Máximo: {max(importes)} €, Mínimo: {min(importes)} €, Ordenados: {sorted(importes)}, Conteo: {len(importes)}")
```

```python
## Ejercicio 8 — Del cálculo con listas al cálculo vectorizado
```

```python
unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios  = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

# 1. Intentar multiplicar listas
try:
    _ = unidades * precios
except TypeError as e:
    print("Error al multiplicar listas directas:", e)
    print("Explicación: El operador '*' no está definido entre dos listas en Python; solo sirve para replicar una lista por un entero.")

# 2. Conversión a NumPy
np_unidades = np.array(unidades)
np_precios = np.array(precios)
print(f"np_unidades - dtype: {np_unidades.dtype}, shape: {np_unidades.shape}")
print(f"np_precios  - dtype: {np_precios.dtype}, shape: {np_precios.shape}")

# 3. Importes vectorizados y suma
importes = np_unidades * np_precios
print("Importes:", importes)
print("Suma total importes:", round(importes.sum(), 2), "€")

# 4. Recargo logístico del 4%
importes_recargo = importes * 1.04
print("Importes con recargo del 4%:", np.round(importes_recargo, 2))

# 5. Array con tipos heterogéneos
arr_mixto = np.array([1, "dos", 3.0, True])
print("Array mixto dtype:", arr_mixto.dtype)
print("Explicación: NumPy fuerza un tipo de dato homogéneo para almacenar los elementos contiguos en memoria y optimizar el rendimiento vectorial en C.")


# 6. Comparación de tiempos de ejecución con %timeit
print("Tiempo con bucle for en Python puro:")
%timeit [u * p for u, p in zip(unidades, precios)]

print("Tiempo vectorizado con NumPy:")
%timeit np_unidades * np_precios
```

```python
## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada
```

```python
# 1. Generar arrays con generador reproducible
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

# 2. Estadísticos de temperatura
media_temp = np.mean(temperatura)
mediana_temp = np.median(temperatura)
std_temp = np.std(temperatura)
min_temp = np.min(temperatura)
max_temp = np.max(temperatura)

print(f"Temperatura -> Media: {media_temp:.2f}, Mediana: {mediana_temp:.2f}, Desv: {std_temp:.2f}, Min: {min_temp:.2f}, Max: {max_temp:.2f}")
print("Razón: La proximidad casi exacta entre media y mediana refleja una distribución prácticamente simétrica, propia de una campana de Gauss normal.")

# 3. Percentiles de vibración
percentiles = np.percentile(vibracion, [25, 50, 75, 95])
print("Percentiles vibración [P25, P50, P75, P95]:", np.round(percentiles, 3))

# 4. Apilar en (500, 2)
telemetria_matriz = np.column_stack((temperatura, vibracion))
print("Forma de telemetria_matriz:", telemetria_matriz.shape)

# 5. Correlación
matriz_corr = np.corrcoef(temperatura, vibracion)
print("Matriz de correlación:\n", np.round(matriz_corr, 4))
print("Explicación: El coeficiente cercano a 0 muestra ausencia de relación lineal entre ambas variables simuladas independientemente.")

# 6. Detección de umbral anómalo
umbral = media_temp + 2 * std_temp
superan_umbral = (temperatura > umbral).sum()
pct_superan = (superan_umbral / len(temperatura)) * 100
print(f"Lecturas que superan el umbral ({umbral:.2f}): {superan_umbral} ({pct_superan:.2f}%)")
```

```python
## Ejercicio 15 — Columnas calculadas, agregación y exportación
```

```python
import os

# 1. Columna importe_bruto
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

# 2. Columna importe_neto con descuento
ventas["importe_neto"] = (ventas["importe_bruto"] * (1 - ventas["descuento_pct"] / 100)).round(2)

# 3. Columna ticket_medio
ventas["ticket_medio"] = (ventas["importe_neto"] / ventas["unidades"]).round(2)

# 4. Agrupación por region y suma de facturación neta ordenada
facturacion_region = ventas.groupby("region")["importe_neto"].sum().sort_values(ascending=False)
print("Facturación neta total por región:\n", facturacion_region)
print("Facturación neta global:", round(ventas["importe_neto"].sum(), 2), "€")

# 5. Agrupación múltiple por categoria y canal
resumen_cat_canal = ventas.groupby(["categoria", "canal"]).agg({
    "importe_neto": "mean",
    "unidades": "sum"
}).round(2)
print("\nResumen por categoría y canal (head 6):\n", resumen_cat_canal.head(6))

# 6. Top 5 productos con mayor importe neto
top5_prod = ventas.groupby("product")["importe_neto"].sum().nlargest(5) if "product" in ventas.columns else ventas.groupby("producto")["importe_neto"].sum().nlargest(5)
print("\nTop 5 productos por facturación:\n", top5_prod)

# 7. Imputación de nulos
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(ventas["satisfaccion"].median())
ventas["canal"] = ventas["canal"].fillna("Desconocido")
print("\nNulos restantes tras imputación:")
print(ventas[["satisfaccion", "canal"]].isna().sum())

# 8. Exportar a outputs
os.makedirs("../outputs", exist_ok=True)
facturacion_region.to_csv("../outputs/facturacion_por_region.csv")
print("\nExportación completada a ../outputs/facturacion_por_region.csv")
```
